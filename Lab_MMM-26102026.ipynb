### 1.2 `generate_container_command`
<a id="misctools-generate_container_command"></a>

Build the command to run a program inside a Docker or Singularity container.

📓 [All the examples in `examples-misctools-methods.ipynb`](https://github.com/connectomicslab/clabtoolkit/blob/main/usage_examples/examples-misctools-methods.ipynb)


```python
######################## Testing generate_container_command ##########################

import clabtoolkit.misctools as cltmisc
import os

print('Testing generate_container_command function with various inputs.')
print('---------------------------------------------------')
# Creating the temporary files
sing_image = '/tmp/my_singularity_image.sif'
docker_image = '/tmp/my_docker_image.tar'
python_script = '/tmp/run_analysis.py'
data_in = '/tmp/input.nii'
data_out = '/tmp/output.nii'
command = ['python', python_script, '--input', data_in, '--output', data_out]

# Generating a fake singularity and docker images for testing
for f in [sing_image, docker_image, python_script, data_in, data_out]:
    with open(f, 'w') as temp_file:
        temp_file.write('# Temporary file for testing\n')

print("Test 1: Generate the container command with Singularity")
container_command = cltmisc.generate_container_command(command, technology='singularity', image_path=sing_image)
print(f'Input Command: {command}')
print(f'Generated Container Command: {container_command}')

# Remove the fake Singularity image after testing
os.remove(sing_image)

print('')

# Remove the temporary files created for the example
for f in [docker_image, python_script, data_in, data_out]:
    os.remove(f)
```

### 3.3 `get_regionwise_timeseries`
<a id="parcellationtools-get_regionwise_timeseries"></a>

Average a 4D image inside each region to get one time series per region.

📓 [All the examples in `examples-parcellationtools-methods.ipynb`](https://github.com/connectomicslab/clabtoolkit/blob/main/usage_examples/examples-parcellationtools-methods.ipynb)


```python
########################### Testing get_regionwise_timeseries ###########################
import clabtoolkit.parcellationtools as cltparc
import clabtoolkit.imagetools as cltimg
import clabtoolkit.visualizationtools as cltvis

# Simulating the parcellation
parc = cltparc.Parcellation.simulate_parcellation(n_regions=10, dimensions=(20, 20, 20), seed=42)
parc.set_names(['ctx-rh-precentral-simu', 'ctx-lh-precentral-simu', 'ctx-rh-postcentral-simu',
                    'ctx-lh-postcentral-simu', 'subc-rh-amygdala-simu', 'subc-lh-amygdala-simu',
                    'hipp-lh-hippocampus-simu', 'hipp-rh-hippocampus-simu', 'cer-lh-cerebellum-simu', 'cer-rh-cerebellum-simu'])

# Simulate a time series data
sim_image = '/tmp/simulated_4d_data.nii.gz'
sim_img = cltimg.simulate_image(
    reference=parc,
    simulated_image=sim_image,
    n_volumes=20
)

# Example 1: Compute region-wise time series from file and removing the first 3 volumes
# NOTE: get_regionwise_timeseries returns a RegionTimeSeries object (not a raw numpy
# array). The actual (n_regions x timepoints) array lives in its `.data` attribute,
# and the region metadata lives in `.region_names` / `.region_colors`.
print("")
print('Test 1: Computing region-wise time series from file...')
region_ts = parc.get_regionwise_timeseries(sim_image, method="clabtoolkit", vols_to_delete="1-3")
print("Region-wise time series shape:", region_ts.data.shape)
# Define groups based on region names (e.g., left and right cortex). The rest will remain ungrouped.
groups = {
                "Left Cortex":    ['ctx-lh-'],
                "Right Cortex":   ['ctx-rh-']
            }
cltvis.create_carpet_plot(region_ts, groups=groups, figsize=(9, 6), tr=2.0, title="Region-wise Time Series Carpet Plot", groups_title="Region Groups")
print("-------------------------------") 
```

### 4.1 `load_scalar_maps`
<a id="surfacetools-load_scalar_maps"></a>

Load vertex-wise maps, such as cortical thickness or curvature, onto a surface.

📓 [All the examples in `examples-surfacetools-methods.ipynb`](https://github.com/connectomicslab/clabtoolkit/blob/main/usage_examples/examples-surfacetools-methods.ipynb)


```python
########################### Testing load_scalar_maps ###########################
import clabtoolkit.surfacetools as cltsurf
import os

############## Loading from a surface file ###################################

# Create a Surface object
freesurfer_home = os.environ.get("FREESURFER_HOME")
if freesurfer_home is None:

    # Create Surface object
    surf_obj  = cltsurf.Surface.simulate_surface(seed=42, shape="sphere") 
    surf_obj.save_surface(filename="/tmp/simulated_surface.surf", overwrite=True)
    ct_map_file = "/tmp/simulated_surface.thickness"
    curv_map_file = "/tmp/simulated_surface.curv"
    surf_file = "/tmp/simulated_surface.surf"

    # Simulating thickness file
    surf_obj.export_map(filename=ct_map_file, overlay_name="thickness", overwrite=True)

    # Simulating curvature file
    surf_obj.simulate_map(map_name="curvature")
    surf_obj.export_map(filename=curv_map_file, overlay_name="curvature", overwrite=True)
    
else:
    subjects_dir = os.path.join(freesurfer_home, 'subjects')
    subject_id = 'fsaverage'
    surf_id = 'pial'
    hemi_id = 'lh'
    surf_file = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.{surf_id}')
    annot_file = os.path.join(subjects_dir, subject_id, 'label', f'{hemi_id}.aparc.annot')
    ct_map_file = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.thickness')
    curv_map_file = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.curv')


# Loading the surface
surf_obj = cltsurf.Surface(surf_file)

# Example 1: Load a morphometry file
print("\nTest 1: Loading the map and use it as overlay")
surf_obj.load_scalar_maps(ct_map_file, maps_names="Thickness") # Assign an ID for the map. It will be used as overlay name
surf_obj.load_scalar_maps(curv_map_file, maps_names="Curvature") # Assign an ID for the map. It will be used as overlay name

surf_obj.get_info()
print('')
surf_obj.plot(maps=["Thickness", "Curvature"], hemi="lh", 
            colorbar_position="right", 
            colorbar_title=["Thickness", "Curvature"],
            cmap=["jet", "BrBG"])
```

### 4.2 `load_annotation`
<a id="surfacetools-load_annotation"></a>

Load a FreeSurfer annotation, with its colors, to show a parcellation on the surface.

### 4.3 `merge_surfaces`
<a id="surfacetools-merge_surfaces"></a>

Merge several surfaces, for example both hemispheres, into a single surface.

📓 [All the examples in `examples-surfacetools-methods.ipynb`](https://github.com/connectomicslab/clabtoolkit/blob/main/usage_examples/examples-surfacetools-methods.ipynb)


```python
########################### Testing merge_surfaces ###########################
import clabtoolkit.surfacetools as cltsurf
import clabtoolkit.freesurfertools as cltfree
import os


freesurfer_home = os.environ.get("FREESURFER_HOME")

subjects_dir = os.path.join(freesurfer_home, 'subjects')
subject_id = 'fsaverage'
surf_id = 'pial'
hemi_id = 'lh'
surf_file_lh = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.{surf_id}')
annot_file_lh = os.path.join(subjects_dir, subject_id, 'label', f'{hemi_id}.aparc.annot')
ct_map_file_lh = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.thickness')
curv_map_file_lh = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.curv')

hemi_id = 'rh'
annot_file_rh = os.path.join(subjects_dir, subject_id, 'label', f'{hemi_id}.aparc.annot')
surf_file_rh = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.{surf_id}')
ct_map_file_rh = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.thickness')
curv_map_file_rh = os.path.join(subjects_dir, subject_id, 'surf', f'{hemi_id}.curv')
parc_name = "Desikan"

maps_names = ["Thickness", "Curvature"]

# Loading the surface and the maps for the left hemisphere
surf_obj_lh = cltsurf.Surface(surf_file_lh)
# Reading multiple FreeSurfer map files
for i, map_file in enumerate([ct_map_file_lh, curv_map_file_lh]):
    surf_obj_lh.load_scalar_maps(map_file, maps_names=maps_names[i])
# Load a surface and an annotation file
surf_obj_lh.load_annotation(annot_file_lh, parc_name=parc_name)

# Loading the surface and the maps for the right hemisphere
surf_obj_rh = cltsurf.Surface(surf_file_rh)
# Reading multiple FreeSurfer map files
for i, map_file in enumerate([ct_map_file_rh, curv_map_file_rh]):
    surf_obj_rh.load_scalar_maps(map_file, maps_names=maps_names[i])
# Load a surface and an annotation file
surf_obj_rh.load_annotation(annot_file_rh, parc_name=parc_name)

# Test 1: Merging the two hemisphere surfaces
print("Test 1: Merging the two hemisphere surfaces...")
# Merge the two hemisphere surfaces into a new surface
merged = cltsurf.merge_surfaces([surf_obj_lh, surf_obj_rh])

merged.plot(maps="Desikan")
```

Keep only the streamlines whose length satisfies a condition, such as `120 < length < 130`.

📓 [All the examples in `examples-tracttools-methods.ipynb`](https://github.com/connectomicslab/clabtoolkit/blob/main/usage_examples/examples-tracttools-methods.ipynb)


```python
######################## Testing filter ###################################
import tempfile
from pathlib import Path
import clabtoolkit.tracttools as clttract
import clabtoolkit.imagetools as cltimg
import clabtoolkit.visualizationtools as cltvis
import copy
import os


freesurfer_home = os.environ.get("FREESURFER_HOME")
if freesurfer_home is None:

    print("FREESURFER_HOME environment variable is not set.")
    print("Simulating reference image and tractogram...")
    simulated_image = str(Path(tempfile.gettempdir()) / "simulated_image.nii.gz")
    ref_image = cltimg.simulate_image(reference=(100, 100, 70), simulated_image=simulated_image)

    # Simulating a tractogram with 5 bundles of length 150 mm
    tract_obj = clttract.Tractogram.simulate_tractogram(ref_image=simulated_image, bundle_length=150)

else:
    in_trk = f"{freesurfer_home}/trctrain/hcp/mni/cc.bodyc.display.trk"
    tract_obj = clttract.Tractogram(in_trk)

lengths = tract_obj.compute_streamline_lengths()

print("Number of streamlines:", len(tract_obj.tracts))
print(f"Streamline lengths between {lengths.min():.2f} and {lengths.max():.2f} mm")

# Test 1: Keeping the streamlines inside a range of lengths.
# filter modifies the object in place, so it is applied to a copy of the tractogram
print("\nTest 1: Filtering the streamlines with lengths between 120 and 130 mm...")
tract_obj_120_130 = copy.deepcopy(tract_obj)
tract_obj_120_130.filter(condition="120 < length < 130")
print("Number of streamlines kept:", len(tract_obj_120_130.tracts))

# Test 2: A single comparison can also be used
print("\nTest 2: Filtering the streamlines longer than 125 mm...")
tract_obj_gt_125 = copy.deepcopy(tract_obj)
tract_obj_gt_125.filter(condition="length > 125")
print("Number of streamlines kept:", len(tract_obj_gt_125.tracts))

# Plotting the original and the filtered tractograms
print("\nVisualizing the original and the filtered tractograms...")
plotter = cltvis.BrainPlotter()
plotter.plot([tract_obj, tract_obj_120_130, tract_obj_gt_125],
             map_names="length",
             colormaps="jet",
             use_opacity=False,
             colorbar_style="shared",
             colorbar_titles="Length (mm)",
             views=["caudal"])
```

### 5.2 `filter_by_mask`
<a id="tracttools-filter_by_mask"></a>

Select the streamlines that touch a mask, or that connect specific pairs of regions of a parcellation.

📓 [All the examples in `examples-tracttools-methods.ipynb`](https://github.com/connectomicslab/clabtoolkit/blob/main/usage_examples/examples-tracttools-methods.ipynb)


```python
######################## Testing filter_by_mask ###################################
import tempfile
from pathlib import Path
import clabtoolkit.tracttools as clttract
import clabtoolkit.imagetools as cltimg
import clabtoolkit.parcellationtools as cltparc
import clabtoolkit.visualizationtools as cltvis
import os


freesurfer_home = os.environ.get("FREESURFER_HOME")
if freesurfer_home is None:

    print("FREESURFER_HOME environment variable is not set.")
    print("Simulating reference image and tractogram...")
    simulated_image = str(Path(tempfile.gettempdir()) / "simulated_image.nii.gz")
    ref_image = cltimg.simulate_image(reference=(100, 100, 70), simulated_image=simulated_image)

    # Simulating a tractogram with 5 bundles of length 150 mm
    tract_obj = clttract.Tractogram.simulate_tractogram(ref_image=simulated_image, bundle_length=150)

    # Simulating a parcellation on the same grid as the tractogram
    parc = cltparc.Parcellation.simulate_parcellation(n_regions=5, dimensions=(100, 100, 70), affine=ref_image.affine, seed=42)
    reg_dict = {1: 5}
else:
    in_trk = f"{freesurfer_home}/trctrain/hcp/mni/cc.bodyc.display.trk"
    tract_obj = clttract.Tractogram(in_trk)
    tract_obj.compute_streamline_lengths()

    subjects_dir = os.path.join(freesurfer_home, 'subjects')
    subject_id = 'bert'
    parc_id = 'aparc+aseg.mgz'
    parc_file = os.path.join(subjects_dir, subject_id, 'mri', f'{parc_id}')
    lut_file = os.path.join(freesurfer_home, 'FreeSurferColorLUT.txt')
    parc = cltparc.Parcellation(parc_file, color_table=lut_file)

    # Left and right precentral cortex
    reg_dict = {1024: 2024}

print("Number of streamlines:", len(tract_obj.tracts))

# Test 1: Keeping the streamlines that connect the pair of regions in reg_dict
print("\nTest 1: Keeping the streamlines that connect the pair of regions")
filtered, bundles, registry = tract_obj.filter_by_mask(parc, reg_dict, split_bundles=True)
print("Number of streamlines kept:", len(filtered.tracts))
for pair_id, info in registry.items():
    print(f"  Pair {pair_id}: {info['name']} ({info['n_streamlines']} streamlines)")

# Plotting the original and the filtered tractograms
print("\nVisualizing the original and the filtered tractograms...")
plotter = cltvis.BrainPlotter()
plotter.plot([tract_obj, filtered],
             map_names="length",
             use_opacity=False,
             colorbar_style="shared",
             colorbar_titles="Length (mm)")
```

```python
######################## Testing get_processing_status_details_json ########################

import clabtoolkit.pipelinetools as cltpipe
import clabtoolkit.bidstools as cltbids
import clabtoolkit.misctools as cltmisc
import os

# Generating a BIDs dataset
bids_dir = "/tmp/sim_bids_dbstatus2"   # <-- Replace with your BIDS root directory
deriv_dir = os.path.join(bids_dir, 'derivatives') # <-- Replace with your derivatives directory path if different

# ------ Simulated BIDS dataset creation (REMOVE this section if using a real dataset) ----- #
# Create a simulated BIDS dataset for testing purposes 
print(f'Creating a simulated BIDS dataset in {bids_dir}...')

nsub = 5 # Number of subjects
nses = 2 # Number of visits
nruns = 1 # Number of runs per modality
modalities = ["anat"] # Define the modalities to include in the simulated dataset

print(f'Number of subjects: {nsub}')
print(f'Number of visits: {nses}')

cltbids.create_a_simulated_bids_dataset(
        output_dir=bids_dir,
        n_subjects=nsub,
        n_visits=nses,
        modalities=modalities,
        include_derivatives=['thalseg-cnn', 'chimera', 'fsl-first'],
        runs_per_modality=nruns,
        overwrite=True,
    )

# Remove 2 files from the simulated chimera dataset and 1 file from the fsl-first dataset to test processing status
os.remove(os.path.join(deriv_dir, 'chimera', 'sub-01', 'ses-01','anat', 'sub-01_ses-01_T1w.nii.gz'))
os.remove(os.path.join(deriv_dir, 'chimera', 'sub-01', 'ses-02','anat', 'sub-01_ses-02_T1w.nii.gz'))
os.remove(os.path.join(deriv_dir, 'fsl-first', 'sub-05', 'ses-02','anat', 'sub-05_ses-02_T1w.nii.gz'))

# ---- End of Simulated BIDS dataset creation (REMOVE this section if using a real dataset) ----- #

# Get the list of Full IDs to process from the BIDs dataset
full_ids = cltpipe.get_fullids2process(in_dir=bids_dir, source='files')

print('\n Test 1: Creating processing status table for simulated BIDs dataset')
output_table = os.path.join(deriv_dir, 'processing_status_table.tsv')
deriv_table, _ = cltpipe.create_processing_status_table(deriv_dir, full_ids=full_ids, output_table=output_table)

processed_subjects, _ = cltpipe.get_processing_status_details_json(
    proc_status_df=output_table,
    full_ids=full_ids,
    deriv_dir=deriv_dir,
    pipe_dirs=["chimera", "fsl-first"], 
    only_ids=False # To get only the full IDs set it to True
)

data_new = cltmisc.ExplorerDict(processed_subjects)

# Print updated tree
print('Updated Dictionary Tree:')
print('The Full Ids containing missing files for the selected pipelines are:')
data_new.tree()
```
